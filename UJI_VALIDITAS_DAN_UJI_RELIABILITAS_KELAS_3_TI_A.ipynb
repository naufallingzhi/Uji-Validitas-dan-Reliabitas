```python
# 1. IMPORT LIBRARY
import re
import pandas as pd
import numpy as np
from scipy.stats import pearsonr, t
from google.colab import files
```

```python
# ============================================================
# 2. UPLOAD DATASET
# ============================================================

uploaded = files.upload()

nama_file = list(uploaded.keys())[0]

df = pd.read_excel(nama_file)

print("Dataset berhasil dibaca!")
print("Jumlah responden:", len(df))
display(df.head())
```

```python
# ============================================================
# 3. PEMISAHAN DATA DEMOGRAFI DAN ITEM
# ============================================================

# Dataset Topik 2 memiliki 2 kolom demografi:
# 1. Cap waktu
# 2. Kelas

JUMLAH_KOLOM_DEMOGRAFI = 2

data_demografi = df.iloc[:, :JUMLAH_KOLOM_DEMOGRAFI]

data_item = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:].copy()

kolom_item = data_item.columns.tolist()

print("Jumlah kolom demografi:", JUMLAH_KOLOM_DEMOGRAFI)
print("Jumlah item pernyataan:", len(kolom_item))
```

```python
# ============================================================
# 4. KONVERSI SKALA LIKERT
# ============================================================

skala_likert = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

data_item = data_item.replace(skala_likert)

# Mengubah data menjadi numerik
data_item = data_item.apply(pd.to_numeric, errors="coerce")

# Memeriksa data kosong
print("\nPemeriksaan data kosong:")
print(data_item.isnull().sum())

if data_item.isnull().sum().sum() > 0:
    raise ValueError(
        "Masih terdapat jawaban kosong atau jawaban yang belum dikenali. "
        "Periksa kembali dataset."
    )

print("\nKonversi skala Likert berhasil!")
```

```python
# ============================================================
# 5. PENGELOMPOKAN ITEM BERDASARKAN DIMENSI
# ============================================================

kelompok_dimensi = {}

for kolom in kolom_item:

    hasil = re.match(r"^(.*?) \[", kolom)

    if hasil:
        nama_dimensi = hasil.group(1)

        if nama_dimensi not in kelompok_dimensi:
            kelompok_dimensi[nama_dimensi] = []

        kelompok_dimensi[nama_dimensi].append(kolom)

print("\nPengelompokan dimensi:")

for dimensi, item in kelompok_dimensi.items():
    print(dimensi, ":", len(item), "item")
```

```python
# ============================================================
# 6. MENGHITUNG R TABEL
# ============================================================

n = len(data_item)

df_tabel = n - 2

alpha_signifikansi = 0.05

t_tabel = t.ppf(
    1 - alpha_signifikansi / 2,
    df_tabel
)

r_tabel = t_tabel / np.sqrt(
    df_tabel + t_tabel**2
)

print("\nHASIL PERHITUNGAN R TABEL")
print("Jumlah responden:", n)
print("Derajat kebebasan:", df_tabel)
print("Taraf signifikansi:", alpha_signifikansi)
print("Nilai r tabel:", round(r_tabel, 3))
```

```python
# ============================================================
# 7. PENGUJIAN VALIDITAS
# ============================================================

hasil_validitas = []

for dimensi, item_dimensi in kelompok_dimensi.items():

    # Menghitung total skor setiap dimensi
    total_skor = data_item[item_dimensi].sum(axis=1)

    for kolom in item_dimensi:

        # Mengambil kode item dari nama kolom
        hasil_kode = re.search(
            r"\[([A-Za-z0-9]+)\.",
            kolom
        )

        kode_item = (
            hasil_kode.group(1)
            if hasil_kode
            else kolom
        )

        # Pearson item-total correlation
        r_hitung, p_value = pearsonr(
            data_item[kolom],
            total_skor
        )

        # Corrected item-total correlation
        total_koreksi = total_skor - data_item[kolom]

        r_corrected, _ = pearsonr(
            data_item[kolom],
            total_koreksi
        )

        # Keputusan validitas berdasarkan r hitung > r tabel
        keputusan = (
            "Valid"
            if r_hitung > r_tabel
            else "Tidak Valid"
        )

        hasil_validitas.append({
            "Dimensi": dimensi,
            "Kode Item": kode_item,
            "Pernyataan": kolom,
            "r Hitung": r_hitung,
            "r Tabel": r_tabel,
            "Corrected Item-Total Correlation": r_corrected,
            "P-Value": p_value,
            "Keputusan": keputusan
        })


tabel_validitas = pd.DataFrame(hasil_validitas)

print("\nHASIL PENGUJIAN VALIDITAS")
display(tabel_validitas)
```

```python
# ============================================================
# 8. FUNGSI CRONBACH'S ALPHA
# ============================================================

def cronbach_alpha(data):

    jumlah_item = data.shape[1]

    if jumlah_item < 2:
        return np.nan

    varians_item = data.var(axis=0, ddof=1)

    total_skor = data.sum(axis=1)

    varians_total = total_skor.var(ddof=1)

    if varians_total == 0:
        return np.nan

    alpha = (
        jumlah_item / (jumlah_item - 1)
    ) * (
        1 - varians_item.sum() / varians_total
    )

    return alpha
```

```python
# ============================================================
# 9. PENGUJIAN RELIABILITAS
# ============================================================

hasil_reliabilitas = []

hasil_alpha_dihapus = []

for dimensi, item_dimensi in kelompok_dimensi.items():

    data_dimensi = data_item[item_dimensi]

    # Menghitung Cronbach's Alpha
    nilai_alpha = cronbach_alpha(data_dimensi)

    # Menentukan kategori reliabilitas
    if pd.isna(nilai_alpha):
        kategori = "Tidak dapat dihitung"
    elif nilai_alpha >= 0.90:
        kategori = "Sangat Baik"
    elif nilai_alpha >= 0.80:
        kategori = "Baik"
    elif nilai_alpha >= 0.70:
        kategori = "Reliabel"
    else:
        kategori = "Tidak Reliabel"

    hasil_reliabilitas.append({
        "Dimensi": dimensi,
        "Jumlah Item": len(item_dimensi),
        "Cronbach's Alpha": nilai_alpha,
        "Keterangan": kategori
    })

    # Alpha jika item dihapus
    # Dihitung hanya jika jumlah item lebih dari 2
    if len(item_dimensi) > 2:

        for kolom in item_dimensi:

            data_tanpa_item = data_dimensi.drop(
                columns=[kolom]
            )

            alpha_dihapus = cronbach_alpha(
                data_tanpa_item
            )

            hasil_kode = re.search(
                r"\[([A-Za-z0-9]+)\.",
                kolom
            )

            kode_item = (
                hasil_kode.group(1)
                if hasil_kode
                else kolom
            )

            hasil_alpha_dihapus.append({
                "Dimensi": dimensi,
                "Kode Item": kode_item,
                "Item yang Dihapus": kolom,
                "Alpha Awal": nilai_alpha,
                "Alpha Jika Item Dihapus": alpha_dihapus
            })


tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)

tabel_alpha_dihapus = pd.DataFrame(hasil_alpha_dihapus)

print("\nHASIL PENGUJIAN RELIABILITAS")
display(tabel_reliabilitas)

print("\nHASIL ALPHA JIKA ITEM DIHAPUS")
display(tabel_alpha_dihapus)
```

```python
# ============================================================
# 10. RINGKASAN HASIL PENGUJIAN
# ============================================================

jumlah_valid = (
    tabel_validitas["Keputusan"] == "Valid"
).sum()

jumlah_tidak_valid = (
    tabel_validitas["Keputusan"] == "Tidak Valid"
).sum()

jumlah_reliabel = (
    tabel_reliabilitas["Cronbach's Alpha"] >= 0.70
).sum()

jumlah_tidak_reliabel = (
    len(tabel_reliabilitas) - jumlah_reliabel
)

print("\n======================================")
print("       RINGKASAN HASIL PENGUJIAN")
print("======================================")

print("Jumlah responden:", n)
print("Jumlah dimensi:", len(kelompok_dimensi))
print("Jumlah item:", len(kolom_item))
print("Nilai r tabel:", round(r_tabel, 3))

print("Item valid:", jumlah_valid)
print("Item tidak valid:", jumlah_tidak_valid)

print("Dimensi reliabel:", jumlah_reliabel)
print("Dimensi tidak reliabel:", jumlah_tidak_reliabel)
```

```python
# ============================================================
# 11. EKSPOR HASIL PENGUJIAN KE EXCEL
# ============================================================

nama_output = "Hasil_Uji_Validitas_Reliabilitas_Topik_2_3TIA.xlsx"

ringkasan = pd.DataFrame({
    "Keterangan": [
        "Jumlah Responden",
        "Jumlah Dimensi",
        "Jumlah Item",
        "r Tabel",
        "Item Valid",
        "Item Tidak Valid",
        "Dimensi Reliabel",
        "Dimensi Tidak Reliabel"
    ],
    "Hasil": [
        n,
        len(kelompok_dimensi),
        len(kolom_item),
        round(r_tabel, 3),
        jumlah_valid,
        jumlah_tidak_valid,
        jumlah_reliabel,
        jumlah_tidak_reliabel
    ]
})

with pd.ExcelWriter(nama_output, engine="openpyxl") as writer:

    data_item.to_excel(
        writer,
        sheet_name="Data_Numerik",
        index=False
    )

    ringkasan.to_excel(
        writer,
        sheet_name="Ringkasan",
        index=False
    )

    tabel_validitas.to_excel(
        writer,
        sheet_name="Validitas",
        index=False
    )

    tabel_reliabilitas.to_excel(
        writer,
        sheet_name="Reliabilitas",
        index=False
    )

    tabel_alpha_dihapus.to_excel(
        writer,
        sheet_name="Alpha_Jika_Dihapus",
        index=False
    )

print("\nFile hasil pengujian berhasil dibuat!")

# Mengunduh file Excel
files.download(nama_output)
```

Perbedaan hasil tersebut terjadi karena pada pengujian r hitung, skor item masih termasuk dalam total skor dimensi yang digunakan untuk menghitung korelasi. Kondisi ini dapat membuat nilai korelasi menjadi lebih tinggi. Sebaliknya, pada Corrected Item-Total Correlation, skor item dikeluarkan terlebih dahulu dari total skor sehingga pengujiannya menjadi lebih ketat. Hasil ini menunjukkan bahwa meskipun seluruh item memenuhi kriteria validitas utama berdasarkan r hitung dan r tabel, sebagian besar item masih memiliki hubungan yang relatif rendah dengan total skor dimensi setelah item tersebut dikeluarkan. Oleh karena itu, item-item tersebut perlu ditinjau kembali dari segi kesesuaian isi dan keterkaitannya dengan dimensi yang diukur.

Sebelum penghapusan, nilai Cronbach's Alpha dimensi Respons Emosional adalah 0,311. Setelah item RE4 dihapus sementara, nilai Cronbach's Alpha meningkat menjadi 0,427 atau naik sebesar 0,116.

Hasil tersebut menunjukkan bahwa item RE4 kemungkinan kurang sesuai dengan item lainnya dalam dimensi Respons Emosional sehingga dapat menurunkan konsistensi internal. Penghapusan RE4 dalam latihan ini merupakan simulasi peninjauan instrumen dan bukan keputusan akhir untuk membuang item, karena seluruh item sebelumnya dinyatakan valid berdasarkan perbandingan r hitung dengan r tabel. Peninjauan isi pernyataan tetap diperlukan sebelum menentukan perbaikan instrumen.

Hal ini menunjukkan bahwa jumlah responden dapat memengaruhi hasil pengujian validitas. Ketika jumlah responden lebih sedikit, nilai r tabel menjadi lebih besar sehingga persyaratan untuk menyatakan suatu butir valid menjadi lebih ketat. Selain itu, perubahan komposisi responden juga dapat memengaruhi nilai korelasi setiap butir pertanyaan. Oleh karena itu, pengujian menggunakan 20 responden ini hanya merupakan simulasi sesuai latihan, sedangkan hasil pengujian awal dengan 39 responden tetap menjadi hasil utama analisis dataset.

5. Refleksi.

Instrumen yang reliabel belum tentu valid karena reliabilitas hanya menunjukkan konsistensi hasil pengukuran, sedangkan validitas menunjukkan ketepatan instrumen dalam mengukur sesuatu yang ingin diukur. Jadi, suatu instrumen bisa memberikan hasil yang konsisten setiap kali digunakan, tetapi belum tentu hasil tersebut sesuai dengan tujuan pengukurannya.

Contohnya adalah sebuah timbangan yang selalu menunjukkan berat badan seseorang lebih berat 3 kg dari berat sebenarnya. Jika digunakan berulang kali, timbangan tersebut tetap memberikan hasil yang sama sehingga dapat dikatakan konsisten atau reliabel. Namun, hasil pengukurannya tidak sesuai dengan berat badan sebenarnya sehingga tidak dapat dikatakan valid.

Dari contoh tersebut, dapat disimpulkan bahwa instrumen yang baik tidak cukup hanya memiliki reliabilitas yang tinggi, tetapi juga harus memiliki validitas yang baik. Keduanya diperlukan agar hasil pengukuran dapat dipercaya dan benar-benar sesuai dengan tujuan penelitian.
